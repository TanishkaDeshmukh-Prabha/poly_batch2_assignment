```python
#Matplotlib assignment-3
# Part A – subplot() (Questions 1–6)

# Q1.
# Create a figure containing 2 subplots:
# ● Line Chart
# ● Bar Chart

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 30, 25, 35, 40]

plt.figure(figsize=(8,4))

# Line Chart
plt.subplot(1,2,1)
plt.plot(months, sales, marker='o')
plt.title("Line Chart")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)

# Bar Chart
plt.subplot(1,2,2)
plt.bar(months, sales)
plt.title("Bar Chart")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Q2.
# Create a figure containing 4 subplots:
# ● Line Chart
# ● Bar Chart
# ● Pie Chart
# ● Scatter Plot
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,15,30,25]

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.plot(x,y,marker='o')
plt.title("Line Chart")
plt.grid(True)

plt.subplot(2,2,2)
plt.bar(x,y)
plt.title("Bar Chart")
plt.grid(True)

plt.subplot(2,2,3)
plt.pie(y, labels=x, autopct='%1.1f%%')
plt.title("Pie Chart")

plt.subplot(2,2,4)
plt.scatter(x,y)
plt.title("Scatter Plot")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Q3.
# Create a 2 × 2 Student Dashboard showing:
# ● Student Marks
# ● Attendance
# ● Subject Distribution
# ● Study Hours vs Marks
import matplotlib.pyplot as plt

subjects = ["Math","Sci","Eng","Comp"]
marks = [85,78,90,88]
attendance = [92,95,89,96]
hours = [2,3,4,5]

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.plot(subjects,marks,marker='o')
plt.title("Student Marks")
plt.grid(True)

plt.subplot(2,2,2)
plt.bar(subjects,attendance)
plt.title("Attendance")
plt.grid(True)

plt.subplot(2,2,3)
plt.pie(marks,labels=subjects,autopct='%1.1f%%')
plt.title("Subject Distribution")

plt.subplot(2,2,4)
plt.scatter(hours,marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Q4.
# Create an Employee Dashboard containing:
# ● Salary Trend
# ● Department-wise Employees
# ● Salary Distribution (Histogram)
# ● Experience vs Salary
import matplotlib.pyplot as plt

experience = [1,2,3,4,5]
salary = [25000,30000,40000,50000,60000]
departments = ["HR","IT","Sales","Finance"]
employees = [8,20,12,10]

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.plot(experience,salary,marker='o')
plt.title("Salary Trend")
plt.grid(True)

plt.subplot(2,2,2)
plt.bar(departments,employees)
plt.title("Department-wise Employees")
plt.grid(True)

plt.subplot(2,2,3)
plt.hist(salary,bins=5)
plt.title("Salary Distribution")
plt.grid(True)

plt.subplot(2,2,4)
plt.scatter(experience,salary)
plt.title("Experience vs Salary")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Q5.
# Create a dashboard with:
# ● Monthly Sales (Line)
# ● Product Sales (Bar)
# ● Payment Methods (Pie)
# ● Customer Ratings (Histogram)
# Use subplot() to display all charts together.
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
sales = [20,25,30,35,40]

products = ["A","B","C","D"]
product_sales = [50,40,30,20]

payment = ["Cash","Card","UPI"]
payment_count = [20,35,45]

ratings = [4,5,3,4,5,5,4,3,2,5]

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.plot(months,sales,marker='o')
plt.title("Monthly Sales")
plt.grid(True)

plt.subplot(2,2,2)
plt.bar(products,product_sales)
plt.title("Product Sales")
plt.grid(True)

plt.subplot(2,2,3)
plt.pie(payment_count,labels=payment,autopct='%1.1f%%')
plt.title("Payment Methods")

plt.subplot(2,2,4)
plt.hist(ratings,bins=5)
plt.title("Customer Ratings")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Q6.
# Create a 3 × 2 subplot layout and display any six different charts learned in Matplotlib.
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[10,20,30,25,35]

plt.figure(figsize=(12,8))

plt.subplot(3,2,1)
plt.plot(x,y)
plt.title("Line")

plt.subplot(3,2,2)
plt.bar(x,y)
plt.title("Bar")

plt.subplot(3,2,3)
plt.scatter(x,y)
plt.title("Scatter")

plt.subplot(3,2,4)
plt.hist(y,bins=5)
plt.title("Histogram")

plt.subplot(3,2,5)
plt.pie(y,labels=x,autopct='%1.1f%%')
plt.title("Pie")

plt.subplot(3,2,6)
plt.boxplot(y)
plt.title("Box Plot")

plt.tight_layout()
plt.show()
```
