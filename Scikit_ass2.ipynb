```python
#Scikit-learn(Assignment-2)

# Question 1: Find Missing Values
# Create the following dataset and display all missing values using:
# ● isnull()
# ● isnull().sum()
# Name  Age  Salary
# Amit  25  30000
# Rahul NaN 45000
# Sneh  28   NaN
# Priya 30 55000
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya"],
    "Age": [25, np.nan, 28, 30],
    "Salary": [30000, 45000, np.nan, 55000]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("\nMissing Values:")
print(df.isnull())

print("\nTotal Missing Values:")
print(df.isnull().sum())
```

```python
# Question 2:
# Remove Missing Values
# Using the dataset from Question 1, remove all rows containing missing values using:
# dropna()
# Display the updated dataset.
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya"],
    "Age": [25, np.nan, 28, 30],
    "Salary": [30000, 45000, np.nan, 55000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

df = df.dropna()

print("\nDataset after removing missing values:")
print(df)
```

```python
# Question 3:
# Replace Missing Values Using Mean
# Create the following dataset:
# Age
# 20
# 25
# NaN
# 35
# 40
# Replace the missing value using the Mean.
import pandas as pd
import numpy as np

data = {
    "Age": [20, 25, np.nan, 35, 40]
}

df = pd.DataFrame(data)

df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

```python
# Question 4:
# Replace Missing Values Using Median
# Create the following dataset:
# Salary
# 30000
# 35000
# 40000
# NaN
# 90000
# Replace the missing value using the Median.
import pandas as pd
import numpy as np

data={
    "Salary":[30000,35000,40000,"NaN",90000]
}

df=pd.DataFrame(data)

df["Salary"] = df["Salary"].fillna(df["Salary"].median())

print(df)
```

```python
# Question 5:
# Replace Missing Values Using Mode
# Create the following dataset:
# City
# Pune
# Mumbai
# Pune
# NaN
# Delhi
# Pune
# Replace the missing value using the Mode.
import pandas as pd
import numpy as np

data={
    "City":["Pune","Mumbai","Pune",np.nan,"Delhi","Pune"]
}
df=pd.DataFrame(data)

df["City"]=df["City"].fillna(df["City"].mode()[0])

print(df)
```

```python
# Question 6:
# Use SimpleImputer
# Create the following dataset:
# Age Salary
# 22 25000
# NaN 35000
# 30 NaN
# 35 50000
# Use SimpleImputer with the Mean strategy to replace missing values
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

data={
    "Age":[22,np.nan,30,35],
    "Salary":[25000,35000,np.nan,50000]
}

df=pd.DataFrame(data)

imputer = SimpleImputer(strategy="mean")
df[["Age","Salary"]]= imputer.fit_transform(df[["Age","Salary"]])

print(df)
```

```python
# Question 7:
# Apply Label Encoding
# Create the following dataset:
# Gender
# Male
# Female
# Male
# Female
# Male
# Apply Label Encoding using LabelEncoder and display the updated dataset.
import pandas as pd
from sklearn.preprocessing import LabelEncoder

data={
    "Gender":["Male","Female","Male","Female","Male"]
}

df=pd.DataFrame(data)

encoder=LabelEncoder()

df["Gender"]=encoder.fit_transform(df["Gender"])

print(df)
```

```python
# Question 8:
# aApply One-Hot Encoding
# Create the following dataset:
# City
# Pune
# Mumbai
# Delhi
# Pune
# Mumbai
# Apply One-Hot Encoding using pd.get_dummies().
# Display the output.
import pandas as pd

data = {
    "City": ["Pune", "Mumbai", "Delhi", "Pune", "Mumbai"]
}

df = pd.DataFrame(data)

df = pd.get_dummies(df)

print(df)
```

```python
# Question 9:
# Complete Data Preprocessing
# Create the following dataset:
# Name Age Salary Gender City
# Amit 25 30000 Male Pune
# Rahul NaN 45000 Male Mumbai
# Sneh
# a
# 28 NaN Female Delhi
# Priya 30 55000 Female Pune
# Perform the following steps:
# 1. Display the dataset.
# 2. Find missing values.
# 3. Replace missing values using SimpleImputer.
# 4. Apply Label Encoding to the Gender column.
# 5. Apply One-Hot Encoding to the City column.
# 6. Display the final processed dataset.
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder

data = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya"],
    "Age": [25, np.nan, 28, 30],
    "Salary": [30000, 45000, np.nan, 55000],
    "Gender": ["Male", "Male", "Female", "Female"],
    "City": ["Pune", "Mumbai", "Delhi", "Pune"]
}
#display dataset
df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

imputer = SimpleImputer(strategy="mean")

df[["Age", "Salary"]] = imputer.fit_transform(df[["Age", "Salary"]])

encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])

df = pd.get_dummies(df, columns=["City"])

print("\nFinal Processed Dataset:")
print(df)
```

```python
# Question 10: 
# Employee Data Preprocessing Challenge
# Create a dataset with the following columns:
# ● Name
# ● Age
# ● Salary
# ● Experience
# ● Gender
# ● Department
# Your dataset should contain:
# ● At least 8 employee records
# ● At least 3 missing values
# ● Two categorical columns (Gender and Department)
# Perform the following operations:
# 1. Display the dataset.
# 2. Find missing values.
# 3. Replace numerical missing values using SimpleImputer (Mean Strategy).
# 4. Apply Label Encoding to the Gender column.
# 5. Apply One-Hot Encoding to the Department column.
# 6. Verify that there are no missing values.
# 7. Display the final processed dataset.
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder

data = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya", "Rohan", "Neha", "Kiran", "Ajay"],
    "Age": [25, np.nan, 28, 30, 35, np.nan, 29, 31],
    "Salary": [30000, 45000, np.nan, 55000, 60000, 48000, np.nan, 65000],
    "Experience": [2, 4, np.nan, 5, 8, 3, 6, 7],
    "Gender": ["Male", "Male", "Female", "Female", "Male", "Female", "Female", "Male"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Finance", "IT", "HR"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

imputer = SimpleImputer(strategy="mean")

df[["Age", "Salary", "Experience"]] = imputer.fit_transform(
    df[["Age", "Salary", "Experience"]]
)

encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])

df = pd.get_dummies(df, columns=["Department"])

print("\nMissing Values After Preprocessing:")
print(df.isnull().sum())

print("\nFinal Processed Dataset:")
print(df)
```
