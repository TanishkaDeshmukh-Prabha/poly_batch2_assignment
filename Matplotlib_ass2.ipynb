```python
# Q25. Final Practical Challenge
# Create a Sales Analysis Dashboard containing:
# ● Product Sales (Bar Chart)
# ● Payment Method (Pie Chart)
# ● Student Marks Distribution (Histogram)
# ● Study Hours vs Marks (Scatter Plot)
# Requirements:
# ● Use plt.subplot() to display all 4 charts in one figure.
# ● Add a main title using plt.suptitle().
# ● Add titles to each chart.
# ● Add labels wherever applicable.
# ● Add grid to Bar Chart, Histogram, and Scatter Plot.
# ● Set figure size to (12, 8).
import matplotlib.pyplot as plt

plt.figure(figsize=(12,8))

# Product Sales - Bar Chart
products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [45000, 30000, 15000, 10000]

plt.subplot(2,2,1)
plt.bar(products, sales, color="green", edgecolor="black")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.grid(axis="y")


# Payment Method - Pie Chart
payment = ["Cash", "UPI", "Card", "Net Banking"]
percentage = [30, 40, 20, 10]

plt.subplot(2,2,2)
plt.pie(percentage, labels=payment, autopct="%1.1f%%", shadow=True)
plt.title("Payment Methods")


# Student Marks Distribution - Histogram
marks = [45, 55, 60, 65, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92, 95]

plt.subplot(2,2,3)
plt.hist(marks, bins=5, color="purple", edgecolor="black")
plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.grid()


# Study Hours vs Marks - Scatter Plot
study_hours = [1, 2, 3, 4, 5, 6, 7]
student_marks = [45, 55, 60, 70, 80, 90, 95]

plt.subplot(2,2,4)
plt.scatter(study_hours, student_marks, color="red", marker="o", s=100)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()


plt.suptitle("Sales Analysis Dashboard")
plt.tight_layout()
plt.show()
```
